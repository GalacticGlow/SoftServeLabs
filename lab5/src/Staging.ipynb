```python
import urllib.request
from datetime import datetime

dbutils.widgets.text("url", "")
dbutils.widgets.text("volume_path", "")

url = dbutils.widgets.get("url")
volume_path = dbutils.widgets.get("volume_path")

current_time = datetime.now().strftime("%Y%m%dT%H%M%SZ")
snapshot_name = f"products_raw_{current_time}.csv.gz"

try:
    urllib.request.urlretrieve(url, f"{volume_path}/{snapshot_name}")
    print(f"File successfully downloaded to {dbutils.widgets.get("volume_path")}")
except Exception as e:
    print(f"An error occurred: {e}")
```
