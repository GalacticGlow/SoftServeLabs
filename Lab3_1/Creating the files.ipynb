##Getting the data from the api

```python
import requests

url = "https://earthquake.usgs.gov/fdsnws/event/1/query"
count = 1000
storage_path = "/Volumes/dbr_dev_ua5816bd/gulidov_artem_staging/raw_csv/earthquakes"

params = {
    "format": "geojson",
    "starttime": "2026-08-01",
    "endtime": "2026-09-01",
    "limit": count,
    "orderby": "time-asc"
}

response = requests.get(url, params=params)
response.raise_for_status()

data = response.json()
```

##Splitting into 1000 json files and uploading to local storage

```python
import json

for i, earthquake in enumerate(data["features"], start=1):

    file_data = {
        "type": "FeatureCollection",
        "features": [earthquake]
    }

    file_path = storage_path + f"/earthquake_{i}.json"

    with open(file_path, "w") as f:
        json.dump(file_data, f)
```
