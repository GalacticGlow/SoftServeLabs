```python
files = {"train.parquet" : "bronze_train", "validation.parquet" : "bronze_validation"}

for file, name in files.items():
    df = spark.read \
        .option("header", True) \
        .option("multiLine", True) \
        .parquet(f"/Volumes/dbr_dev_ua5816bd/gulidov_artem_staging/raw_csv/{file}")

    (
        df.write
        .format("delta")
        .mode("overwrite")
        .option("mergeSchema", "true")
        .saveAsTable(f"dbr_dev_ua5816bd.gulidov_artem_bronze.{name}")
    )
```
