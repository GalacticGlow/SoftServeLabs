```python
import urllib.request

url = 'https://static.openfoodfacts.org/data/en.openfoodfacts.org.products.csv.gz'
volume_path = "/Volumes/dbr_dev_ua5816bd/gulidov_artem_staging/raw_csv/products_raw.csv.gz"
adls_raw_path = "abfss://artgulidov@dlsua5816bd.dfs.core.windows.net/staging/products_raw"

try:
    urllib.request.urlretrieve(url, volume_path)
    print(f"File successfully downloaded to {volume_path}")
except Exception as e:
    print(f"An error occurred: {e}")
```

```python
df = spark.read \
    .option("header", True) \
    .option("delimiter", "\t") \
    .option("multiLine", True) \
    .option("escape", "\"") \
    .option("compression", "gzip") \
    .csv(volume_path)

(
    df.write
    .format("delta")
    .mode("overwrite")
    .option("mergeSchema", "true")
    .option("path", adls_raw_path)
    .saveAsTable("dbr_dev_ua5816bd.gulidov_artem_staging.off_staging")
)
```
