```python
import pyspark.sql.functions as F

df = spark.read \
    .option("header", True) \
    .option("delimiter", "\t") \
    .option("multiLine", True) \
    .option("escape", "\"") \
    .option("compression", "gzip") \
    .csv(VOLUME_PATH)

df = df.withColumn("ingested_at", F.current_timestamp())

(
    df.write
    .format("delta")
    .mode("append")
    .option("mergeSchema", "true")
    .saveAsTable(f"{CATALOG}.{BRONZE_SCHEMA}.{BRONZE_TABLE}")
)
```
